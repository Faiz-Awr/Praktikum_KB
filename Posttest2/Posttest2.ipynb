## Visualisasi data menggunakan grafik batang untuk kategori/label 'satisfied'/'neutral or dissatisfied'

Berdasarkan grafik batang tersebut, kesimpulan utamanya adalah jumlah penumpang yang merasa netral atau tidak puas lebih banyak dibandingkan jumlah penumpang yang merasa puas.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(3,4))

sns.countplot(x='satisfaction', hue='satisfaction', data=df, palette='Set3', legend=False)

plt.title("Tingkat Kepuasan Penumpang")
plt.xlabel("Satisfaction")
plt.ylabel("Jumlah Penumpang")

plt.show()
```

## Menampilkan korelasi dataset menggunakan heatmap.

#### 1. Korelasi Terkuat: Keterlambatan Keberangkatan dan Kedatangan

Terdapat korelasi positif yang sangat kuat (0.96) antara Departure Delay in Minutes dan Arrival Delay in Minutes. Artinya jika pesawat mengalami keterlambatan saat berangkat, hampir pasti pesawat tersebut juga akan terlambat saat arrival.

#### 2. Korelasi Sedang: Layanan Dalam Penerbangan (In-flight Service Cluster)

Cleanliness (Kebersihan) memiliki korelasi kuat dengan Seat comfort (0.68), Inflight entertainment (0.7), dan Food and drink (0.66). Seat comfort (Kenyamanan Kursi) juga berkorelasi kuat dengan Inflight entertainment (0.62). Artinya penumpang yang merasa puas dengan satu aspek layanan di dalam kabin (misalnya kebersihan), cenderung akan merasa puas juga dengan aspek lainnya (seperti kenyamanan kursi atau hiburan).


```python
corr = df.corr()
plt.figure(figsize=(20,10))
sns.heatmap(corr, cmap='coolwarm', annot=True, center=0)
plt.title("Heatmap Korelasi Antar Atribut")
plt.show()
```
