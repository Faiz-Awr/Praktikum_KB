Melakukan pemeriksaan untuk kolom yang memiliki missing value

## Feature Engineering
Membuat beberapa fitur baru

## 1. Membuat kategori umur (Age_Group)

kepuasan penumpang mungkin berbeda antar kelompok umur. Anak-anak mungkin lebih mudah terhibur, sementara penumpang dewasa lebih peduli pada kenyamanan dan ketepatan waktu

```python
bins = [0, 12, 18, 35, 60, 100] 
labels = ['Child', 'Teen', 'Young Adult', 'Adult', 'Senior']
df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels)

print(df[['Age', 'Age_Group']].head())
```

## 2. Membuat Kategori Jarak Penerbangan (Distance_Category) 


```python
bins = [0, 1000, 3000, 6000] 
labels = ['Short', 'Medium', 'Long']
df['Distance_Category'] = pd.cut(df['Flight Distance'], bins=bins, labels=labels)

print(df[['Flight Distance', 'Distance_Category']].head())
```

## 3. Membuat Skor Total Layanan (Total_Service_Score)

```python
service_cols = ['Inflight wifi service', 'Departure/Arrival time convenient', 
                'Ease of Online booking', 'Gate location', 'Food and drink', 
                'Online boarding', 'Seat comfort', 'Inflight entertainment', 
                'On-board service', 'Leg room service', 'Baggage handling', 
                'Checkin service', 'Inflight service', 'Cleanliness']

# Jumlahkan semua skor layanan
df['Total_Service_Score'] = df[service_cols].sum(axis=1)

# Lihat hasilnya
print(df[['Total_Service_Score', 'satisfaction']].head())
```

## 4. Membuat Fitur (Is_Delayed) 
informasi delay menjadi ya (1) atau tidak (0)

```python
df['Is_Delayed'] = ((df['Departure Delay in Minutes'] > 0) | (df['Arrival Delay in Minutes'] > 0)).astype(int)

print(df[['Departure Delay in Minutes', 'Arrival Delay in Minutes', 'Is_Delayed']].head())
```

## Melakukan Scalling untuk kolom numerik dan Melakukan encoding untuk kolom kategorikal

mengubah dua nilai unik menjadi angka 0 dan 1. Contohnya, {'Female': 0, 'Male': 1} atau {'satisfied': 1, 'neutral or dissatisfied': 0}. 
untuk kolom Class (Eco, Business, Eco Plus) akan menggunakan metode one-hot encoding

Kolom yang perlu di-Scaling: Age, Flight Distance, Departure Delay Log & Arrival Delay Log dan semua kolom penilaian layanan (dari Inflight wifi service hingga Cleanliness)


```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, LabelEncoder
# --- Encoding ---
# One-Hot Encoding untuk kolom kategorikal dengan banyak label
# Ini akan mengubah 'Class', 'Age_Group', dan 'Distance_Category' menjadi angka 0 dan 1
df = pd.get_dummies(df, columns=['Class', 'Age_Group', 'Distance_Category'], drop_first=True)

#Label Encoding
le = LabelEncoder()
binary_cols = ['Gender', 'Customer Type', 'Type of Travel']
for col in binary_cols:
    df[col] = le.fit_transform(df[col])

# --- Menghapus kolom yang tidak diperlukan lagi ---
# buang kolom asli yang sudah dibuat fiturnya atau tidak relevan
columns_to_drop = [
    'Unnamed: 0', 
    'id', 
    'Age',                  # Sudah diwakili oleh Age_Group
    'Flight Distance',      # Sudah diwakili oleh Distance_Category
    'Departure Delay in Minutes', # Sudah diwakili oleh Log dan Is_Delayed
    'Arrival Delay in Minutes'    # Sudah diwakili oleh Log dan Is_Delayed
]
df = df.drop(columns=columns_to_drop)

# 'satisfaction' adalah target yang ingin kita prediksi
X = df.drop(columns=['satisfaction'])
y = df['satisfaction']

# Ubah target 'satisfaction' menjadi 0 dan 1 juga
y = le.fit_transform(y)

# --- Scaling ---
# Terapkan StandardScaler ke semua kolom fitur (X)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Membuat DataFrame dari hasil scaling untuk kemudahan melihat hasilnya
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

print("Berikut adalah 5 baris pertama dari data fitur (X) yang siap digunakan:")
print(X_scaled_df.head())

```

## Split Data

Split menjadi train 80% dan test 20%

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled_df, 
    y, 
    test_size=0.2, 
    random_state=42
)

#Melihat Dimensi Data X, y Train dan Test
print('Dimensi X train: ', X_train.shape)
print('Dimensi X test: ', X_test.shape)
print('Dimensi y train: ', y_train.shape)
print('Dimensi y test: ', y_test.shape)
```

## Implementasi Metode Support Vector Classifier

model SVC dilatih untuk mengklasifikasikan penumpang ke dalam dua kategori: 'satisfied' dan 'neutral or not satisfied'. Jadi model akan belajar dari data train yang berjumlah 80% dari total row di dataset.

```python
import pandas as pd
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt

# --- 1. Latih model ---
print("Membuat model Support Vector Classifier...")
model_svc = SVC(random_state=42)
print("Model sedang berlatih...")
model_svc.fit(X_train, y_train)
print("Model telah selesai berlatih")
```

### Model melakukan Prediksi

Jadi model disini akan melakukan prediksi untuk 20% data yang belum pernah dilihatnya sebagai bukti pengujian bahwa model ini bisa diandalkan atau akurat. Berdasarkan outputnya terlihat bahwa accuracynya berada di 94% yang berarti performanya bagus. 

Untuk precisionnya Ketika model memprediksi seorang penumpang 'satisfied' (label 1), tebakannya benar sebanyak 94%. Begitu pula saat menebak 'neutral or dissastified' (label 0), tebakannya juga benar 94%. 

Pada recall, model berhasil mengidentifikasi 95% dari semua penumpang yang sebenarnya 'neutral or dissastified' dan 91% dari semua penumpang yang sebenarnya 'satisfied'.

Skor F1 seimbang (0.94 dan 0.92) menunjukkan bahwa model memiliki keseimbangan yang baik antara presisi dan recall untuk kedua kelas.

```python
# --- 3. Model memprediski ---
print("Membuat prediksi pada data testing...")
y_pred_svc = model_svc.predict(X_test)


# --- 4. Evaluasi Performa Model ---
print("\n--- Hasil Evaluasi Model SVC ---")
# Menampilkan laporan klasifikasi yang lengkap
print(classification_report(y_test, y_pred_svc))
```

### Confusion Matrix
Kesimpulan dari confusion matrixnya adalah sebagai berikut.

True Negative (Kiri Atas): Model dengan benar memprediksi 2804 penumpang sebagai 'neutral or dissastified'.
False Positive (Kanan Atas): Model membuat kesalahan dengan memprediksi 141 penumpang sebagai 'satisfied', padahal sebenarnya mereka neutral or dissastified.

True Positive (Kanan Bawah): Model dengan benar memprediksi 2058 penumpang sebagai 'satisfied'.
False Negative (Kiri Bawah): Model membuat kesalahan dengan "melewatkan" 193 penumpang yang sebenarnya 'satisfied', dan salah memprediksinya sebagai neutral or dissastified

```python
# --- 5. Visualisasi Confusion Matrix ---
# Menampilkan matriks untuk melihat di mana model benar dan salah
print("Menampilkan Confusion Matrix...")
cm = confusion_matrix(y_test, y_pred_svc)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Neutral or dissastified', 'Satisfied'])
disp.plot(cmap=plt.cm.Blues)
plt.title("Confusion Matrix")
plt.show()
```
